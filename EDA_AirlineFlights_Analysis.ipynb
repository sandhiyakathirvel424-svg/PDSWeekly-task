TASK 2 : UNIVARIATE ANALYSIS

```python
import pandas as pd
import matplotlib.pyplot as plt

dataframe = pd.read_csv("airlines_flights_data.csv.xls")

dataframe["price"].hist()
plt.show()

dataframe["airline"].value_counts().plot(kind="bar")
plt.show()

dataframe["class"].value_counts().plot(kind="bar")
plt.show()
```

TASK 3: BIVARIATE ANALYSIS

```python
import seaborn as sns

sns.boxplot(x="class", y="price",data=dataframe)
plt.show()

sns.scatterplot(x="duration",y="price", data=dataframe)
plt.show()
```

TASK 4: CORRELATION & FEATURE ENGINEERING

```python
dataframe["class"] = dataframe["class"].astype("category").cat.codes

sns.heatmap(dataframe.corr(numeric_only=True), annot=True)
plt.show()
```

```python
dataframe["price_per_hour"] = dataframe["price"] / dataframe["duration"]
dataframe["is_direct"] = (dataframe["stops"] == 0).astype(int)
```

TASK 5: BUSINESS INSIGHTS

```python
import pandas as pd

# Load dataset
dataframe = pd.read_csv("airlines_flights_data.csv.xls")

# Top 5 Business Insights
print("1. Average Ticket Price by Airline")
print(dataframe.groupby("airline")["price"].mean())

print("\n2. Average Ticket Price by Class")
print(dataframe.groupby("class")["price"].mean())

print("\n3. Average Ticket Price by Stops")
print(dataframe.groupby("stops")["price"].mean())

print("\n4. Average Ticket Price by Duration")
print(dataframe.groupby("duration")["price"].mean())

print("\n5. Total Flights by Airline")
print(dataframe["airline"].value_counts())

# Recommendations
print("\nRecommendations:")
print("1. Give discounts during low-demand periods.")
print("2. Use dynamic pricing based on demand.")
print("3. Encourage early booking with special offers.")
```
