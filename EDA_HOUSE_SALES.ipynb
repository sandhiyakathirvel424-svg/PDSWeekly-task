```python
data = df.groupby("bedrooms")["price"].mean()
data.plot(kind="bar")
plt.title("House Prices")
plt.xlabel("Bedrooms")
plt.ylabel("Price")
plt.show()
```

```python
df["date"] = pd.to_datetime(df["date"])
data = df.groupby(df["date"].dt.year)["price"].mean()
data.plot(kind="line", marker="o")
plt.title("Price Trend")
plt.xlabel("Year")
plt.ylabel("Price")
plt.show()
```

```python
data = df["floors"].value_counts()
data.plot(kind="pie", autopct="%1.1f%%")
plt.title("House Floors")
plt.ylabel("")
plt.show()
```

```python
plt.scatter(df["sqft_living"], df["price"])
plt.title("Living Area vs Price")
plt.xlabel("Living Area")
plt.ylabel("Price")
plt.show()
```

```python
data = pd.crosstab(df["floors"], df["condition"])
sns.heatmap(data, annot=True, cmap="coolwarm")
plt.title("Floors vs Condition")
plt.show()
```
